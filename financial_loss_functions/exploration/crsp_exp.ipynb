### 3. Cross-Sectional View (Global View):
#### Cross-sectional mean and vol of returns

```python
ret_cols = [c for c in crsp_predictors.columns if c.endswith('_RET')]
crsp_predictors['cs_mean_ret'] = crsp_predictors[ret_cols].mean(axis=1)
crsp_predictors['cs_std_ret'] = crsp_predictors[ret_cols].std(axis=1)

ax = crsp_predictors.set_index('date')[['cs_mean_ret','cs_std_ret']].plot(figsize=(12,4), title='Cross-Sectional Mean & Vol of Returns')

plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

#### Cross-sectional percentiles / regime visualization
Shows how the cross-sectional spread changes (e.g., crisis widening)

```python
cs = crsp_predictors[ret_cols]
pcts = cs.quantile([0.05, 0.25, 0.5, 0.75, 0.95], axis=1).T
pcts.index = crsp_predictors['date']

fig, ax = plt.subplots(1,1,figsize=(12,4))
ax.fill_between(pcts.index, pcts[0.05], pcts[0.95], alpha=0.2, label='5-95%')
ax.fill_between(pcts.index, pcts[0.25], pcts[0.75], alpha=0.3, label='25-75%')
ax.plot(pcts.index, pcts[0.5], lw=1, label='median')
ax.set_title('Cross-section: median and percentile bands of returns')
ax.legend(); plt.tight_layout(); plt.show()
```

### 4. Feature Distribution Analysis

#### Q–Q plot and histogram grid for multiple features (pooled across tickers)

```python
features = ['_RET','_VOL_CHANGE','_TURNOVER','_ILLIQUIDITY', '_BA_SPREAD']  # suffixes
# build pooled series for each feature: stack across tickers and dropna
pooled = {}
for suf in features:
    cols = [c for c in crsp_predictors.columns if c.endswith(suf)]
    pooled[suf] = crsp_predictors[cols].stack().dropna().values

# plot grid: histogram + Q-Q per feature
n = len(features)
fig, axes = plt.subplots(n, 2, figsize=(10, 4*n))
for i, suf in enumerate(features):
    ax_hist = axes[i,0]
    ax_qq   = axes[i,1]
    ax_hist.hist(pooled[suf], bins=100, density=True)
    ax_hist.set_title(f'{suf} pooled hist')
    stats.probplot(pooled[suf], dist="norm", plot=ax_qq)
    ax_qq.set_title(f'{suf} Q-Q vs Normal')
plt.tight_layout(); plt.show()
```

#### Formal tests (Anderson-Darling, KS)

```python
results = []
for suf in features:
    arr = pooled[suf]
    ad = stats.anderson(arr, dist='norm')  # returns statistic and critical values
    ks = stats.kstest((arr - arr.mean())/arr.std(), 'norm')
    results.append((suf, ad.statistic, ks.statistic, ks.pvalue))
pd.DataFrame(results, columns=['feature','AD_stat','KS_stat','KS_p']).set_index('feature')
```

High AD/KS stats & low p indicate departure from normality.

#### Normal Distribution vs Student-t Q-Q Plot Test

```python
crsp_predictors_t = 5  # degrees of freedom for Student-t

fig, axes = plt.subplots(len(features), 2, figsize=(10,4*len(features)))

for i, suf in enumerate(features):
    data = pooled[suf]

    # Q-Q vs Normal
    sm.qqplot(data, dist=stats.norm, line="s", ax=axes[i,0])
    axes[i,0].set_title(f"{suf} vs Normal")

    # Q-Q vs Student-t  (note: pass distribution object, not a string)
    sm.qqplot(data, dist=t, distargs=(crsp_predictors_t,), line="s", ax=axes[i,1])
    axes[i,1].set_title(f"{suf} vs Student-t (crsp_predictors={crsp_predictors_t})")

plt.tight_layout()
plt.show()
```

#### Detect changing distributions over time

```python
crsp_predictors['year'] = crsp_predictors['date'].dt.year
years = sorted(crsp_predictors['year'].unique())
base_year = years[0]
w_dist = {suf: [] for suf in features}

base_data = {suf: crsp_predictors[crsp_predictors['year']==base_year][[c for c in crsp_predictors.columns if c.endswith(suf)]].stack().dropna().values for suf in features}

for y in years:
    for suf in features:
        ydata = crsp_predictors[crsp_predictors['year']==y][[c for c in crsp_predictors.columns if c.endswith(suf)]].stack().dropna().values
        w = stats.wasserstein_distance(base_data[suf], ydata) if len(ydata)>0 else np.nan
        w_dist[suf].append(w)

# plot
plt.figure(figsize=(10,4))
for suf in features:
    plt.plot(years, w_dist[suf], marker='o', label=suf)
plt.xlabel('year'); plt.ylabel('Wasserstein distance from year '+str(base_year)); plt.legend(); plt.title('Distribution drift by year')
plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

#### Testing for distributions for each feature:
#### a) Returns:

```python
# --- Safe fit helpers ---
def safe_fit(dist, data):
    try:
        return dist.fit(data)
    except Exception:
        return None

def safe_ks(dist, params, data):
    try:
        D, p = stats.kstest(data, dist.name, args=params)
        return D, p
    except Exception:
        return None, None

# --- Main function ---
def fit_and_compare_distributions(df, suffix, bins=200):
    cols = [c for c in df.columns if c.endswith(suffix)]
    data = df[cols].stack().dropna().astype(float).values

    print(f"Feature: {suffix}, Observations: {len(data):,}\n")

    distributions = {
        "Normal": stats.norm,
        "Laplace": stats.laplace,
        "StudentT": stats.t,
        "Lognormal": stats.lognorm,
        "Gamma": stats.gamma,
    }

    results = []
    data = np.array(data)

    for name, dist in distributions.items():

        # skip lognormal & gamma if non-positive
        if name in ["Lognormal", "Gamma"] and np.any(data <= 0):
            continue

        params = safe_fit(dist, data)
        if params is None:
            continue

        D, p = safe_ks(dist, params, data)
        if D is None:
            continue
        
        # log-likelihood
        try:
            ll = np.sum(dist.logpdf(data, *params))
            if not np.isfinite(ll):
                continue
        except Exception:
            continue

        k = len(params)
        AIC = 2*k - 2*ll
        results.append([name, D, p, AIC, params])

    results_df = pd.DataFrame(
        results, columns=["Distribution","KS Stat","KS p-value","AIC","Params"]
    ).sort_values("AIC")

    print(results_df[["Distribution","KS Stat","KS p-value","AIC"]], "\n")

    # Best fit
    best = results_df.iloc[0]
    best_dist = distributions[best["Distribution"]]
    frozen = best_dist(*best["Params"])

    # --- Plotting ---
    plt.figure(figsize=(15,4))

    # Histogram + PDF
    plt.subplot(1,3,1)
    plt.hist(data, bins=bins, density=True)
    x = np.linspace(data.min(), data.max(), 1000)
    plt.plot(x, frozen.pdf(x))
    plt.title(f"{suffix}: Histogram + {best['Distribution']} PDF")

    # QQ Plot
    plt.subplot(1,3,2)
    
    dist_name = best["Distribution"]
    params = best["Params"]
    
    # Handle distributions with shape parameters
    if dist_name == "StudentT":
        df_param = params[0]
        osm, osr = stats.probplot(data, dist="t", sparams=(df_param,), fit=False)
    
    elif dist_name == "Lognormal":
        shape = params[0]
        osm, osr = stats.probplot(data, dist="lognorm", sparams=(shape,), fit=False)
    
    elif dist_name == "Gamma":
        shape = params[0]
        osm, osr = stats.probplot(data, dist="gamma", sparams=(shape,), fit=False)
    
    else:
        # Normal / Laplace etc.
        osm, osr = stats.probplot(data, dist=best_dist.name, fit=False)
    
    plt.scatter(osm, osr, s=3)
    slope, intercept, *_ = linregress(osm, osr)
    xx = np.linspace(min(osm), max(osm), 200)
    plt.plot(xx, intercept + slope*xx, '--')
    plt.title(f"{suffix}: QQ Plot ({dist_name})")



    # CDF comparison
    plt.subplot(1,3,3)
    sorted_data = np.sort(data)
    ecdf = np.arange(len(data))/len(data)
    plt.plot(sorted_data, ecdf, label="Empirical")
    plt.plot(sorted_data, frozen.cdf(sorted_data), label=f"{best['Distribution']} CDF")
    plt.legend()
    plt.title(f"{suffix}: CDF Comparison")
    
    plt.tight_layout()
    plt.show()

    return results_df
```

```python
fit_and_compare_distributions(crsp_predictors, "_RET")
```

`Inference: Returns seem to follow Student-t distribution`
It has heavy tails and is not normal.
